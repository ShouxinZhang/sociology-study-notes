# 硬币分布实验

模拟 $N$ 次公平硬币投掷，并计算正面朝上的频率。公平硬币的正面与反面概率均为 $0.5$。

## 实验函数

```python
import random


def simulate_coin_flips(n: int = 10_000) -> tuple[int, float]:
    """模拟 n 次公平硬币投掷，返回正面次数和正面频率。"""
    if isinstance(n, bool) or not isinstance(n, int) or n <= 0:
        raise ValueError("投掷次数 n 必须是正整数")

    heads_count = sum(random.random() < 0.5 for _ in range(n))
    heads_frequency = heads_count / n
    return heads_count, heads_frequency
```

## 实验参数

修改 `N` 可以调整硬币投掷次数；默认投掷 10,000 次。

```python
N = 10**6
```

## 运行实验

```python
random.seed(6)
heads_count, heads_frequency = simulate_coin_flips(N)

print(f"投掷次数：{N}")
print(f"正面次数：{heads_count}")
print(f"正面频率：{heads_frequency:.6f}（{heads_frequency:.2%}）")
```
